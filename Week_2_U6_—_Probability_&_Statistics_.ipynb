```python
import seaborn as sns
import matplotlib.pyplot as plt


df = sns.load_dataset("tips")

tip = df["tip"]

mean_tip = tip.mean()
median_tip = tip.median()
mode_tip = tip.mode()[0]

print("Mean:", mean_tip)
print("Median:", median_tip)
print("Mode:", mode_tip)

print("Mean > Median:", mean_tip > median_tip)

plt.hist(tip, bins=10, edgecolor="black")

plt.axvline(mean_tip, linestyle="--", label="Mean")
plt.axvline(median_tip, linestyle="-", label="Median")

plt.xlabel("Tip")
plt.ylabel("Frequency")
plt.title("Distribution of Tips")
plt.legend()
plt.show()
```

```python
import seaborn as sns


df = sns.load_dataset("tips")

tip = df["tip"]

tip_range = tip.max() - tip.min()
tip_variance = tip.var()
tip_std = tip.std()

print("Range:", tip_range)
print("Variance:", tip_variance)
print("Standard deviation:", tip_std)

Q1 = tip.quantile(0.25)
Q3 = tip.quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Most robust measure of spread: IQR")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")
tip = df["tip"]

five_number = {
    "Minimum": tip.min(),
    "Q1": tip.quantile(0.25),
    "Median": tip.median(),
    "Q3": tip.quantile(0.75),
    "Maximum": tip.max()
}

print("Five-number summary:")
print(five_number)

skewness = tip.skew()
print("\nSkewness:", skewness)

if skewness > 0:
    print("Comment: The data is positively skewed (leans to the right).")
elif skewness < 0:
    print("Comment: The data is negatively skewed (leans to the left).")
else:
    print("Comment: The data is approximately symmetric.")

# 3. Box plot of tip
plt.figure(figsize=(7, 4))
sns.boxplot(x=tip)

plt.title("Box Plot of Tips")
plt.xlabel("Tip")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

corr_matrix = df.select_dtypes(include="number").corr().round(2)

print("Correlation Matrix:")
print(corr_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()

print("\nDescriptive Statistics:")
print(df.describe())
```
