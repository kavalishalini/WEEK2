```python
import numpy as np

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),  # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),  # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0]))   # pair 3
]

for i, (u, v) in enumerate(pairs, start=1):
    # 1. L1 and L2 norm of u
    l1_norm = np.linalg.norm(u, ord=1)
    l2_norm = np.linalg.norm(u, ord=2)


    cosine_similarity = np.dot(u, v) / (
        np.linalg.norm(u) * np.linalg.norm(v)
    )

    print(f"Pair {i}:")
    print("L1 norm of u:", l1_norm)
    print("L2 norm of u:", l2_norm)
    print("Cosine similarity:", cosine_similarity)
    print()
```

```python
import numpy as np

tri = np.array([
    [0, 1, 0.5],
    [0, 0, 1.0]
])


S = np.array([
    [2, 0],
    [0, 0.5]
])

scaled_tri = S @ tri

print("Scaling matrix:\n", S)
print("Scaled triangle:\n", scaled_tri)
```

```python
import numpy as np


angle = np.radians(45)

R = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle),  np.cos(angle)]
])

rotated_tri = R @ tri

print("Rotation matrix:\n", R)
print("Rotated triangle:\n", rotated_tri)

```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([
    [0, 1, 0.5],
    [0, 0, 1.0]
])

S = np.array([
    [2, 0],
    [0, 0.5]
])


angle = np.radians(45)
R = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle),  np.cos(angle)]
])


transformed_tri = R @ S @ tri


plt.figure(figsize=(7, 6))
plt.plot(
    np.append(tri[0], tri[0, 0]),
    np.append(tri[1], tri[1, 0]),
    marker="o",
    label="Original"
)

plt.plot(
    np.append(transformed_tri[0], transformed_tri[0, 0]),
    np.append(transformed_tri[1], transformed_tri[1, 0]),
    marker="o",
    label="Transformed"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Original vs Transformed Triangle")
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

```python
import numpy as np

C = np.array([
    [4., 1.],
    [2., 3.]
])

eigenvalues, eigenvectors = np.linalg.eig(C)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
print("Eigenvalues:", eigenvalues)
```

```python

v = eigenvectors[:, 0]
lam = eigenvalues[0]

Av = C @ v
lambda_v = lam * v

print("A v:", Av)
print("lambda v:", lambda_v)
print("Verification:", np.allclose(Av, lambda_v))
```

```python
import numpy as np

A2 = np.array([
    [3., 2.],
    [1., 4.]
])

b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])


x2 = np.linalg.solve(A2, b2)
print("Solution x:", x2)

rank_A2 = np.linalg.matrix_rank(A2)
print("Rank of A2:", rank_A2)
```

```python
cos_cat_dog = np.dot(cat, dog) / (
    np.linalg.norm(cat) * np.linalg.norm(dog)
)

cos_cat_car = np.dot(cat, car) / (
    np.linalg.norm(cat) * np.linalg.norm(car)
)

print("Cosine(cat, dog):", cos_cat_dog)
print("Cosine(cat, car):", cos_cat_car)

# Find which is more similar
if cos_cat_dog > cos_cat_car:
    print("Cat and dog are more similar.")
else:
    print("Cat and car are more similar.")
```
