```python
def g(x):
    return x**3 + 2*x

x = 2
h = 1e-6

derivative = (g(x + h) - g(x)) / h

print("Numerical derivative at x = 2:", derivative)
```

```python
import sympy as sp

x = sp.symbols("x")
f = x**3 + 2*x

symbolic_derivative = sp.diff(f, x)
print("Symbolic derivative:", symbolic_derivative)

symbolic_value = symbolic_derivative.subs(x, 2)
print("Symbolic derivative at x = 2:", symbolic_value)


print("Numerical derivative at x = 2:", derivative)
```

```python
import sympy as sp

x, y = sp.symbols("x y")

h2 = x**2 * y + sp.sin(y)

dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx:", dh_dx)
print("dh/dy:", dh_dy)

gradient = [dh_dx, dh_dy]

print("Gradient:", gradient)
```

```python
gradient_value = [expr.subs({x: 2, y: 0}) for expr in gradient]

print("Gradient at (2, 0):", gradient_value)
```

```python
import sympy as sp


x = sp.symbols("x")

y = sp.exp(x**2 + 1)
inner = x**2 + 1
inner_derivative = 2*x
outer_derivative = sp.exp(inner)

derivative = outer_derivative * inner_derivative

print("Derivative:", derivative)
```

```python
derivative_sympy = sp.diff(y, x)
print("Derivative using sp.diff:", derivative_sympy)

print("Do they match?", sp.simplify(derivative - derivative_sympy) == 0)
```

```python
import numpy as np

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)

Wa = np.random.randn(4, 8) * 0.1   # input -> hidden
Wb = np.random.randn(8, 1) * 0.1   # hidden -> output

z1 = Xb @ Wa

h = np.maximum(0, z1)


y_hat = h @ Wb

loss = np.mean((y_hat - Yb) ** 2)

print("z1:\n", z1)
print("h (ReLU):\n", h)
print("y_hat:\n", y_hat)
print("Loss:", loss)
```

```python
n = Xb.shape[0]

dy = (2 / n) * (y_hat - Yb)

dWb = h.T @ dy


dh = dy @ Wb.T

dz1 = dh * (z1 > 0)

dWa = Xb.T @ dz1

print("dy shape:", dy.shape)
print("dWb shape:", dWb.shape)
print("dh shape:", dh.shape)
print("dz1 shape:", dz1.shape)
print("dWa shape:", dWa.shape)



lr = 0.05

loss_before = loss

Wa = Wa - lr * dWa
Wb = Wb - lr *
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb

loss_after = np.mean((y_hat_new - Yb) ** 2)

print("Loss before:", loss_before)
print("Loss after:", loss_after)
```
